```python

import pandas as pd
from sklearn.linear_model import LogisticRegression


def train_and_evaluate(
    train_hours,
    train_pass,
    new_hours,
    scholarship_threshold=0.7
):
    # Step 1: Build the training DataFrame
    df = pd.DataFrame({
        "hours": train_hours,
        "pass": train_pass
    })

    # Input feature and target
    X = df[["hours"]]
    y = df["pass"]

    # Step 2: Create and train the Logistic Regression model
    logistic_reg_model = LogisticRegression()
    logistic_reg_model.fit(X, y)

    # Step 3: Prepare the new students' data
    new_df = pd.DataFrame({
        "hours": new_hours
    })

    # Select probability of class 1: Pass
    probabilities = logistic_reg_model.predict_proba(new_df)[:, 1]

    print("\nProbabilities:", probabilities)

    # Step 4: Apply default 0.5 threshold
    predicted_class = (probabilities >= 0.5).astype(int)

    # Step 5: Apply scholarship threshold
    scholarship_eligible = probabilities >= scholarship_threshold

    # Step 6: Create the result DataFrame
    result_df = pd.DataFrame({
        "hours": new_hours,
        "probability": probabilities.round(2),
        "predicted_class": predicted_class,
        "scholarship_eligible": scholarship_eligible
    })

    return result_df


if __name__ == "__main__":
    train_hours = [1, 2, 3, 4, 5, 6, 7]
    train_pass = [0, 0, 0, 1, 1, 1, 1]
    new_hours = [2, 3.5, 4, 5.5, 8]

    result = train_and_evaluate(
        train_hours,
        train_pass,
        new_hours
    )

    print("\nFinal result:")
    print(result)
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression

def train_and_evaluate(train_hours, train_pass, new_hours, scholarship_threshold=0.7):
    # Step 1: build the training DataFrame
    train_df = pd.DataFrame({'hours': train_hours, 'pass': train_pass})

    # Step 2: create and fit the LogisticRegression object
    clf = LogisticRegression()
    clf.fit(train_df[['hours']], train_df['pass'])

    # Step 3: predict probability of passing for the new hours values
    new_df = pd.DataFrame({'hours': new_hours})
    proba_pass = clf.predict_proba(new_df[['hours']])[:, 1]

    # Step 4: default 0.5 threshold -> standard predicted class
    predicted_class = np.where(proba_pass >= 0.5, 'pass', 'fail')

    # Step 5: stricter threshold -> scholarship eligibility flag
    scholarship_eligible = np.where(proba_pass >= scholarship_threshold, 'Yes', 'No')

    # Step 6: assemble the summary
    result = pd.DataFrame({
        'hours': new_hours,
        'probability_pass': np.round(proba_pass, 2),
        'predicted_class': predicted_class,
        'scholarship_eligible': scholarship_eligible
    })
    return result

if __name__ == "__main__":
    train_hours = [1, 2, 3, 4, 5, 6, 7]
    train_pass = [0, 0, 0, 1, 1, 1, 1]
    new_hours = [2, 3.5, 4, 5.5, 8]
    print(train_and_evaluate(train_hours, train_pass, new_hours))
```
