```python
import pandas as pd
import numpy as np
from io import StringIO

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Step 1: Load the inlined data into a DataFrame
data = """experience,salary
1,28000
2,36000
3,44000
4,52000
5,60000
6,68000
7,76000
8,84000
9,92000
10,100000
"""

df = pd.read_csv(StringIO(data))
# df.describe()

X = df[["experience"]]
Y = df["salary"]

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size = 0.2,
    random_state = 42
)

model = LinearRegression()
model.fit(X_train, Y_train)
y_pred = model.predict(X_test)

mae = mean_absolute_error(Y_test, y_predict)
mse = mean_squared_error(Y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y_test, y_pred)

# Step 7: Print model parameters
# print("Slope (Coefficient):", model.coef_[0])
# print("Intercept:", model.intercept_)

# # Step 8: Print evaluation metrics
# print("\nModel Evaluation")
# print("MAE :", mae)
# print("MSE :", mse)
# print("RMSE:", rmse)
# print("R²  :", r2)

# print("Y Test :" , Y_test)
# print("\nY Predi :" , y_pred)


# new_candidate = pd.DataFrame({'experience' :[40]})

# newcandidate_sal = model.predict(new_candidate)

# print("\nNew Candidate Salary : ", newcandidate_sal[0])


```
