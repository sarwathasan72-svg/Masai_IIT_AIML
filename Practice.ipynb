```python
# Box plot
import numpy as np
import matplotlib.pyplot as plt

sales = np.array([20, 30, 40, 50, 40, 80, 150, 90, 10, -200])

# Range
data_range = sales.max() - sales.min()
print("Range =", data_range)

# Quartiles
Q1 = np.percentile(sales, 25)
Q3 = np.percentile(sales, 75)

print("Q1 =", Q1)
print("Q3 =", Q3)

# IQR
IQR = Q3 - Q1
print("IQR =", IQR)

# Boundaries
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower Boundary =", lower)
print("Upper Boundary =", upper)

# Outliers
outliers = sales[(sales < lower) | (sales > upper)]
print("Outliers =", outliers)

# Boxplot
plt.boxplot(sales)
plt.show()
```
