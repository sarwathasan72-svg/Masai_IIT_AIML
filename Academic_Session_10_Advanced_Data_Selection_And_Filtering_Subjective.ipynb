```python
# Filter to keep only employees whose salary is strictly above 50,000 and whose experience is strictly greater than 2 years.

import pandas as pd
import numpy as np

data = {
    'employee':   ['Amit',  'Maya', 'Alex',    'Kiran', 'Sarah',     'Kumar'],
    'department': ['HR',    'IT',   'Finance', 'IT',    'Marketing', 'HR'],
    'salary':     [45000,   70000,  65000,     72000,   50000,       48000],
    'experience': [2,       5,      2,         6,       3,           1]
}


df= pd.DataFrame(data)
print("Actual data:" , df)

filtered_data = (df['salary'] > 50000) & (df['experience'] > 2)

print(" Filtered data :", df[filtered_data])

#Print filtered data
#print("Filtered Data :" df(filtered_data))
```
