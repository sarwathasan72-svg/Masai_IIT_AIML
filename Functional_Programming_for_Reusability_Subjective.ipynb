```python
MONTHLY_TAX_THRESHOLD = 600000 / 12

def count_taxed(employees):
    # your implementation HERE
    cnt_taxed_employees = 0
    for name, salary in employees:
        if salary > MONTHLY_TAX_THRESHOLD:
            cnt_taxed_employees += 1
    return cnt_taxed_employees
    pass

def total_salary(salaries):
    # your implementation here
    total = 0
    for salary in salaries:
        total += salary
    return total
    pass

def average_salary(salaries):
    # your implementation here — must call total_salary internally
    total = total_salary(salaries)


    return total // len(salaries)
    pass

# Sample data
employees = [
    ("Alice",   80000),
    ("Bob",    200000),
    ("Carol",   45000),
    ("Dave",   120000),
    ("Eve",     30000),
]
salaries = [80000, 200000, 45000, 120000, 30000]

if __name__ == "__main__":
    print("Taxed employees:", count_taxed(employees))
    print("Total salary:", total_salary(salaries))
    print("Average salary:", average_salary(salaries))
```
