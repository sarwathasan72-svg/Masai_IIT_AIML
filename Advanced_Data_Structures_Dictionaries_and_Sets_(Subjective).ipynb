```python
# Function to convert raw record into list of Dict

#Converts a list of such flat student records into a list of dictionaries, each #with keys ""name"", and a nested ""marks"" dictionary with keys ""math"", #""english"", ""physics"", and ""chemistry"".

raw_records = [
    ["Alice", 80, 75, 90, 85],
    ["Bob", 55, 60, 70, 65],
    ["Carol", 90, 88, 92, 87]
    ]

def convert_records(raw_records):

  if not raw_records:
    return []

#convert raw data in dict

  list_of_dict = []

  for record in raw_records:

    dict_record = {
      "name" :record[0],
        "marks" : {
        "Maths" : record[1],
        "English" : record[2],
        "Physics" :record[3],
        "Chemistry" : record[4]
        }
    }
    list_of_dict.append(dict_record)

  return list_of_dict


converted_records = convert_records(raw_records)

for record in converted_records:
  print(record)


# def students_above_threshold(records, threshold):

print("Calling Threashhold function")

def students_above_threshold(records, threshold):

  if not records:
    return []

  passed_students = set()
  for record in records:
    name = record.get("name", {})
    marks = record.get("marks", {})
    #if all(subject >= threshold for subject in marks.values()):
    #  passed_students.add(name)
    math_marks = marks.get("Maths", 0)
    english_marks = marks.get("English", 0)
    physics_marks = marks.get("Physics", 0)
    chemistry_marks = marks.get("Chemistry", 0)
    if math_marks >= threshold and english_marks >= threshold and physics_marks >= threshold and chemistry_marks >= threshold:
      passed_students.add(name)
  return passed_students

print(students_above_threshold(converted_records, 70))
print(students_above_threshold(converted_records, 85))

set_above_70 = students_above_threshold(converted_records, 70)
set_above_85 = students_above_threshold(converted_records, 85)


print("Symmetric Difference:", set_above_70.symmetric_difference(set_above_85))



```
