```python

import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split


data = pd.DataFrame({
    "area_sqft":   [650, 800, 950, 1100, 700, 1250, 900, 1400, 750, 1000, 1150, 850],
    "bedrooms":    [1, 2, 2, 3, 1, 3, 2, 3, 2, 2, 3, 2],
    "distance_km": [3.5, 2.0, 4.5, 1.5, 5.0, 2.5, 3.0, 1.0, 6.0, 2.8, 3.5, 4.0],
    "rent_k":      [18, 25, 24, 34, 16, 38, 26, 42, 19, 29, 33, 23]
})

x = data.drop("rent_k", axis =1)
y = data["rent_k"]


#  Split train and Test 75% and 25% data
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size = 0.25,
    random_state = 42
)

#  create intance of model & train using features
model = LinearRegression()
model.fit(x_train, y_train)

print("Intercept: ", model.intercept_)
print("Coefficient:", model.coef_)

# Based on Coefficient bedrooms has largest positive effect

# Prediction on test
prediction = model.predict(x_test)
print("Prediction on test: ", prediction)
y_test

# MAE
mae = mean_squared_error(y_test,prediction)
print("mae:", mae)

# MSE
mae = mean_squared_error(y_test,prediction)
print("mae:", mae)

# RMSE
rmse = root_mean_squared_error(y_test,prediction)
print("rmse:", rmse)

#R^^2
r2 = r2_score(y_test, prediction)

print("r:", r2)




```
