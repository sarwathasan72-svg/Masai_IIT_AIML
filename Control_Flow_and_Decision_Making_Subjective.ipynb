```python
#A backend system tracks user events. Events arrive as a list of raw #dictionaries containing user_id, action, and age. The team needs to write a #function that safely filters these events based on complex rules.

raw_events = [
    {"user_id": 1, "action": "login", "age": 25},
    {"user_id": 2, "action": "hack", "age": 30},
    {"user_id": 3, "action": "purchase"},  # missing age
    {"user_id": 4, "action": "logout", "age": 150}, # invalid age
    {"user_id": 5, "action": "purchase", "age": 45}
]

def filter_valid_events(events):
    valid_events = []
    for event in events:
        v_action= event.get("action","Unknown")
        v_age = event.get("age", -1)
        if (v_age < 0 or v_age > 120) or (v_action not in ["login", "logout", "purchase"]):
            continue
        else:

            valid_events.append(event)
    return valid_events

print(filter_valid_events(raw_events))
```
